## 🚗 Car Auction Price Prediction

Given *data about cars for sale in an auction*, let's try to predict the **price** of a given car. 

We will use various regression models to make our predictions.

Data source: https://www.kaggle.com/datasets/doaaalsenani/usa-cers-dataset/code

### Importing Libraries

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import re
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.svm import LinearSVR, SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
```

```python
data = pd.read_csv('archive/USA_cars_datasets.csv')
data
```

```python
data.info()
```

### Preprocessing

```python
df = data.copy()
```

```python
df
```

```python
# Drop Unnamed:0 column
df = df.drop(['Unnamed: 0'], axis=1)
```

```python
{column: df[column].unique() for column in df.select_dtypes('object').columns}
```

```python
{column: len(df[column].unique()) for column in df.select_dtypes('object').columns}
```

```python
len(df['lot'].unique())
```

```python
# Drop high cardinality columns
df = df.drop(['vin', 'lot'], axis=1)
```

```python
def binary_encode(df, columns_with_positive_values):
    df = df.copy()
    for column, positive_value in columns_with_positive_values:
        df[column] = df[column].apply(lambda x: 1 if x == positive_value else 0)
    return df

def onehot_encode(df, columns_with_prefixes):
    df = df.copy()
    for column, prefix in columns_with_prefixes:
        dummies = pd.get_dummies(df[column], dtype=int, prefix=prefix)
        df = pd.concat([df, dummies], axis=1)
        df = df.drop(column, axis=1)
    return df
```

```python
df['country'].unique()
```

```python
df['title_status'].unique()
```

```python
# Binary Encoding
columns_with_positive_values = [('title_status', 'clean vehicle'), ('country', ' usa')]
df = binary_encode(df, columns_with_positive_values)
df
```

```python
# Nominal encoding
columns_with_prefixes = [('brand', 'B'), ('model', 'M'), ('color', 'C'), ('state', 'S'), ('condition', 'C')]
df = onehot_encode(df, columns_with_prefixes)
```

```python
df
```

```python
# Fixing lightgbm column name error occuring later while training
df = df.rename(columns=lambda x: re.sub('[^A-Za-z0-9_]+', '', x))
```

```python
# Split df into X and y
y = df['price'].copy()
X = df.drop('price', axis=1).copy()
```

```python
y
```

```python
X
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, shuffle=True, random_state=123)
```

```python
X_train
```

```python
# Scale X with StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)

X_train = pd.DataFrame(scaler.transform(X_train), columns=X_train.columns)
X_test = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)
```

```python
X_train
```

```python
y_train
```

### Training

```python
models = {
    "                     Linear Regression": LinearRegression(),
    "                   K-Nearest Neighbors": KNeighborsRegressor(),
    "                        Neural Network": MLPRegressor(),
    "Support Vector Machine (Linear Kernel)": LinearSVR(),
    "   Support Vector Machine (RBF Kernel)": SVR(),
    "                         Decision Tree": DecisionTreeRegressor(),
    "                         Random Forest": RandomForestRegressor(),
    "                     Gradient Boosting": GradientBoostingRegressor(),
    "                               XGBoost": XGBRegressor(),
    "                              LightGBM": LGBMRegressor(),
    "                              CatBoost": CatBoostRegressor(verbose=0)
}
```

```python
for name, model in models.items():
    model.fit(X_train, y_train)
    print(name + " trained.")
```

### Results

```python
for name, model in models.items():
    print(name + " R^2 Score: {:.5f}".format(model.score(X_test, y_test)))
```
