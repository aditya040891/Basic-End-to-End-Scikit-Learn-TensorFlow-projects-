## 🍔 NYC Restaurant Result Prediction

Given *data about restaurants in New York*, let's try to predict whether a given restaurant will be marked with a **critical flag**.

We will use a logistic regression model to make our predictions.

Data source: https://www.kaggle.com/datasets/new-york-city/nyc-inspections

### Importing Libraries

```python
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', None)

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import confusion_matrix, classification_report
```

```python
data = pd.read_csv('archive/DOHMH_New_York_City_Restaurant_Inspection_Results.csv')
data
```

```python
data.info()
```

### Preprocessing

```python
df = data.copy()
```

```python
# Drop CAMIS and DBA columns
df = df.drop(['CAMIS', 'DBA'], axis=1)
```

```python
df
```

```python
# Drop unnecessary (duplicate) columns
df = df.drop(['VIOLATION DESCRIPTION'], axis=1)
```

```python
{column: len(df[column].unique()) for column in df.columns}
```

```python
# Drop single valued column
df = df.drop('RECORD DATE', axis=1)
```

```python
df.isna().mean()
```

```python
df.columns[df.isna().mean() > 0.25]
```

```python
# Drop columns with more than 25% missing values
df = df.drop(df.columns[df.isna().mean() > 0.25], axis=1)
```

```python
{column: len(df[column].unique()) for column in df.columns}
```

```python
# Drop high cardinality columns
high_cardinality_columns = ['BUILDING', 'STREET', 'PHONE']
df = df.drop(high_cardinality_columns, axis=1)
```

```python
{column: len(df[column].unique()) for column in df.columns}
```

```python
df
```

```python
# Convert ZIPCODE column into a string column
df['ZIPCODE'] = df['ZIPCODE'].astype(str)
```

```python
df
```

```python
df['CRITICAL FLAG'].unique()
```

```python
# Drop rows with missing target values
missing_target_rows = df.loc[df['CRITICAL FLAG'] == 'Not Applicable'].index
missing_target_rows
```

```python
df = df.drop(missing_target_rows, axis=0).reset_index(drop=True)
```

```python
df
```

```python
df.isna().sum()
```

```python
# Fill missing values 
df['SCORE'] = df['SCORE'].fillna(df['SCORE'].mean())
```

```python
df.isna().sum()
```

```python
# Date encoding
df['INSPECTION DATE'] = pd.to_datetime(df['INSPECTION DATE'])
```

```python
df['INSPECTION YEAR'] = df['INSPECTION DATE'].dt.year
df['INSPECTION MONTH'] = df['INSPECTION DATE'].dt.month
df['INSPECTION DAY'] = df['INSPECTION DATE'].dt.day
df = df.drop('INSPECTION DATE', axis=1)
```

```python
df
```

```python
# One Hot Encode remaining categorical columns
for column in df.select_dtypes('object').columns.drop('CRITICAL FLAG'):
    dummies = pd.get_dummies(df[column], prefix=column)
    df = pd.concat([df, dummies], axis=1)
    df = df.drop(column, axis=1)
```

```python
df
```

### Splitting and Scaling

```python
y = df['CRITICAL FLAG'].copy()
X = df.drop('CRITICAL FLAG', axis=1).copy()
```

```python
y
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, shuffle=True, random_state=42)
```

```python
# Scale X
scaler = StandardScaler()
scaler.fit(X_train)
X_train = pd.DataFrame(scaler.transform(X_train), columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)
```

```python
y_train
```

```python
X_train
```

### Training

```python
y_train.unique()
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

### Results

```python
acc = model.score(X_test, y_test)
print("Test Accuracy: {:.2f}%".format(acc * 100))
```

```python
# Confusion Matrix and Classification Report
y_pred = model.predict(X_test)

cm = confusion_matrix(y_test, y_pred, labels=['Not Critical', 'Critical'])

clr = classification_report(y_test, y_pred, target_names=['Not Critical', 'Critical'])

plt.figure(figsize=(6, 6))
sns.heatmap(cm, annot=True, vmin=0, fmt='g', cmap='Blues', cbar=False)
plt.xticks(ticks=np.arange(2) + 0.5, labels=['Not Critical', 'Critical'])
plt.yticks(ticks=np.arange(2) + 0.5, labels=['Not Critical', 'Critical'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

print("Classification Report:\n---------------------------------------\n", clr)
```

#### Feature Importance

```python
# Use weights of the model to see importance of the features
most_important_feature = np.argmax(model.coef_)
most_important_feature
```
