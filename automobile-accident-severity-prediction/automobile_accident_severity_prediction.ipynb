## 🚗 Automobile Accident Severity Prediction

Given *data about accidents in the US*, let's try to predict the **severity** of a given accident.

We will use a TensorFlow ANN to make our predictions.

Data source: https://www.kaggle.com/datasets/sobhanmoosavi/us-accidents

### Importing Libraries

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

import tensorflow as tf

pd.set_option('display.max_columns', None)
```

```python
data = pd.read_csv('archive/US_Accidents_March23.csv', nrows=100000)
```

```python
data
```

```python
data.info()
```

### Preprocessing

```python
df = data.copy()
```

```python
df.isna().mean()
```

```python
null_columns = ['End_Lat', 'End_Lng', 'Precipitation(in)', 'Wind_Chill(F)']

df = df.drop(null_columns, axis=1)
```

```python
df.isna().sum()
```

```python
df = df.dropna(axis=0).reset_index(drop=True)
```

```python
print("Total missing values:", df.isna().sum().sum())
```

```python
df
```

```python
# Remove unnecessary columns
{column: len(df[column].unique()) for column in df.columns if df.dtypes[column] == 'str'}
```

```python
high_cardinality_columns = ['Zipcode', 'ID', 'Description', 'Street']

df = df.drop(high_cardinality_columns, axis=1)
```

```python
{column: len(df[column].unique()) for column in df.columns if df.dtypes[column] == 'str'}
```

```python
# Removing Country with only one value
df = df.drop('Country', axis=1)
```

```python
df
```

```python
def get_years(df, column):
    return df[column].apply(lambda x: x[0:4])
    
def get_months(df, column):
    return df[column].apply(lambda x: x[5:7])
```

```python
df['Start_Time_Year'] = get_years(df, 'Start_Time').astype(int)
df['End_Time_Year'] = get_years(df, 'End_Time').astype(int)
df['Weather_Timestamp_Year'] = get_years(df, 'Weather_Timestamp').astype(int)

df['Start_Time_Month'] = get_months(df, 'Start_Time').astype(int)
df['End_Time_Month'] = get_months(df, 'End_Time').astype(int)
df['Weather_Timestamp_Month'] = get_months(df, 'Weather_Timestamp').astype(int)

df = df.drop(['End_Time', 'Start_Time', 'Weather_Timestamp'], axis=1)
```

```python
df
```

#### Encoding 

```python
def onehot_encode(df, columns, prefixes):
    df = df.copy()
    for column, prefix in zip(columns, prefixes):
        dummies = pd.get_dummies(df[column], prefix=prefix)
        df = pd.concat([df, dummies], axis=1)
        df = df.drop(column, axis=1)
    return df
```

```python
{column: len(df[column].unique()) for column in df.columns if df.dtypes[column] == 'str'}
```

```python
df = onehot_encode(
    df,
    ['City', 'County', 'State', 'Airport_Code', 'Wind_Direction', 'Weather_Condition'],
    ['CI', 'CO', 'ST', 'AC', 'WD', 'WC']
)
```

```python
df
```

```python
{column: len(df[column].unique()) for column in df.columns if df.dtypes[column] == 'str'}
```

```python
def get_binary_column(df, column):
    df = df.copy()
    if column == 'Source':
        return df[column].apply(lambda x: 1 if x == 'MapQuest' else 0)
    if column == 'Timezone':
        return df[column].apply(lambda x: 1 if x == 'US/Eastern' else 0)
    else:
        return df[column].apply(lambda x: 1 if x == 'Day' else 0)
```

```python
df['Sunrise_Sunset'] = get_binary_column(df, 'Sunrise_Sunset')
df['Civil_Twilight'] = get_binary_column(df, 'Civil_Twilight')
df['Nautical_Twilight'] = get_binary_column(df, 'Nautical_Twilight')
df['Astronomical_Twilight'] = get_binary_column(df, 'Astronomical_Twilight')
df['Source'] = get_binary_column(df, 'Source')
df['Timezone'] = get_binary_column(df, 'Timezone')
```

```python
df
```

#### Splitting and Scaling

```python
y = df['Severity'].copy()
X = df.drop('Severity', axis=1).copy()
```

```python
y.unique()
```

```python
y = y -1
```

```python
y.unique()
```

```python
scaler = StandardScaler()

X = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=1)
```

### Training

```python
X.shape
```

```python
inputs = tf.keras.Input(shape=(X.shape[1], ))
x = tf.keras.layers.Dense(64, activation='relu')(inputs)
x = tf.keras.layers.Dense(64, activation='relu')(x)
outputs = tf.keras.layers.Dense(4, activation='softmax')(x)

model = tf.keras.Model(inputs, outputs)
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

batch_size = 32
epochs = 20

history = model.fit(
    X_train,
    y_train,
    validation_split = 0.2,
    batch_size=batch_size,
    epochs=epochs,
    callbacks=[
        tf.keras.callbacks.ReduceLROnPlateau(),
        tf.keras.callbacks.EarlyStopping(
            monitor='val_loss',
            patience=3,
            restore_best_weights=True
        )
    ]
)
```

### Results

```python
print("Test Accuracy:", model.evaluate(X_test, y_test)[1])
```
