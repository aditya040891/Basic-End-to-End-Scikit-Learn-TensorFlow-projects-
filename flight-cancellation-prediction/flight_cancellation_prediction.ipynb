## Flight Cancellation Prediction

Given *data about US flights in 2015*, let's try to predict whether a given flight will be **cancelled**.

We will use a logistic regression model to make our predictions.

Data source: https://www.kaggle.com/datasets/usdot/flight-delays?resource=download

### Importing Libraries

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import confusion_matrix, classification_report

pd.set_option('display.max_columns', None)
```

```python
data = pd.read_csv('archive/flights.csv', nrows=50000)
data
```

```python
data.info()
```

### Preprocessing

```python
df = data.copy()
```

```python
df.isna().sum()
```

```python
missing_columns = df.loc[:, df.isna().mean() >= 0.25].columns
missing_columns
```

```python
# Remove columns with more than 25% missing values
df = df.drop(missing_columns, axis=1)
df
```

```python
{column: len(df[column].unique()) for column in df.columns}
```

```python
# Drop unneeded columns
df = df.drop(['YEAR', 'MONTH', 'FLIGHT_NUMBER', 'TAIL_NUMBER'], axis=1)
```

```python
def onehot_encode(df, column_dict):
    df = df.copy()
    for column, prefix in column_dict.items():
        dummies = pd.get_dummies(df[column], prefix=prefix)
        df = pd.concat([df, dummies], axis=1)
        df = df.drop(column, axis=1)
    return df
```

```python
# One-hot encode nominal feature columns
df = onehot_encode(df, 
                   column_dict={
                       'AIRLINE': 'AL',
                       'ORIGIN_AIRPORT': 'OA',
                       'DESTINATION_AIRPORT': 'DA'
                   })
```

```python
df
```

```python
missing_value_columns = df.loc[:, df.isna().sum() > 0].columns
missing_value_columns
```

```python
# Fill remaining missing values with column means
for column in missing_value_columns:
    df[column] = df[column].fillna(df[column].mean())
```

```python
df.isna().sum().sum()
```

```python
# Split df into X and y
y = df['CANCELLED'].copy()
X = df.drop('CANCELLED', axis=1).copy()
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=123)
```

```python
# Scale X with a standard scaler
scaler = StandardScaler()
scaler.fit(X_train)

X_train = pd.DataFrame(scaler.transform(X_train), columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)
```

```python
X_train
```

```python
(X_train.mean()).astype(int) == 0
```

### Training

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

### Results

```python
def evaluate_model(model, X_test, y_test):
    model_acc = model.score(X_test, y_test)
    print("Test Accuracy: {:.2f}%".format(model_acc*100))

    y_true = np.array(y_test)
    y_pred = model.predict(X_test)

    cm = confusion_matrix(y_true, y_pred)
    clr = classification_report(y_true, y_pred, target_names=['Not Cancelled', 'Cancelled'])

    plt.figure(figsize=(6, 6))
    sns.heatmap(cm, annot=True, vmin=0, fmt='g', cmap='Blues', cbar=False)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.xticks(np.arange(2)+0.5, labels=['Not Cancelled', 'Cancelled'])
    plt.yticks(np.arange(2)+0.5, labels=['Not Cancelled', 'Cancelled'])
    plt.title("Confusion Matrix")
    plt.show()

    print("Classification Report:\n----------------------------------\n", clr)
```

```python
evaluate_model(model, X_test, y_test)
```
