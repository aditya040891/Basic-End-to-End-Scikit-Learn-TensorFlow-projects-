## Predicting University Rankings

Data source: https://www.kaggle.com/datasets/mylesoneill/world-university-rankings?resource=download

### Importing Libraries

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('archive/cwurData.csv')
data
```

```python
data = data.drop('institution', axis=1)
```

```python
data
```

```python
data = data.drop('year', axis=1)
```

```python
data
```

```python
data.isna().sum()
```

```python
data = data.drop('broad_impact', axis=1)
```

```python
data
```

```python
encoder = LabelEncoder()

data['country'] = encoder.fit_transform(data['country'])

country_mappings = {index: label for index, label in enumerate(encoder.classes_)}
country_mappings
```

```python
data
```

```python
y = data['world_rank'].copy()
X = data.drop('world_rank', axis=1).copy()
```

```python
y
```

```python
X
```

```python
scaler = StandardScaler()

X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=1)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

### Results

```python
print(f"Model R2 Score: {model.score(X_test, y_test):.4f}")
```
